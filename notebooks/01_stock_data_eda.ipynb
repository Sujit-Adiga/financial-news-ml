```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Removing rows with NaN values in the index column

raw_dir = "../data/raw/"
companies = ["RELIANCE", "HDFCBANK", "TCS", "INFY"]

for company in companies:
    file_path = raw_dir + f"{company}_stock_data.csv"
    df = pd.read_csv(file_path)
    df = df[df["index"].notna()]
    df = df.rename(columns={"index": "Date"})
    df["Date"] = pd.to_datetime(df["Date"])
    df.to_csv(file_path, index=False)
    print(f"Cleaned {file_path}")

```

# Plotting Closing prices of the 4 datasets

```python
tcs_df = pd.read_csv("../data/raw/TCS_stock_data.csv")
reliance_df = pd.read_csv("../data/raw/RELIANCE_stock_data.csv")
infy_df = pd.read_csv("../data/raw/INFY_stock_data.csv")
hdfcbank_df = pd.read_csv("../data/raw/HDFCBANK_stock_data.csv")

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(tcs_df["Date"], tcs_df["Close"], label="TCS")
ax.plot(reliance_df["Date"], reliance_df["Close"], label="RELIANCE")
ax.plot(infy_df["Date"], infy_df["Close"], label="INFY")
ax.plot(hdfcbank_df["Date"], hdfcbank_df["Close"], label="HDFCBANK")

ax.set_xlabel("Date")
ax.set_ylabel("Closing Price")
ax.set_title("Stock Closing Prices Over Time")

ax.xaxis.set_major_locator(plt.MaxNLocator(10))
ax.tick_params(axis="x", rotation=45)

ax.legend()
plt.tight_layout()
plt.show()
```
