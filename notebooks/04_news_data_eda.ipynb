```python
import pandas as pd

# Load CNBC news data
cnbc_df = pd.read_csv("../data/raw/headline_data/cnbc_headlines.csv")
cnbc_df.head(10)
```

```python
print(f"CNBC dataset shape: {cnbc_df.shape}")
print(f"CNBC dataset columns: {cnbc_df.columns.tolist()}")
print(f"CNBC dataset missing values:\n{cnbc_df.isnull().sum()}")
```

```python
# Drop NaN from CNBC dataset
cnbc_df.dropna(inplace=True)
cnbc_df.head(10)
```

```python
print(f"CNBC dataset shape: {cnbc_df.shape}")
print(f"CNBC dataset columns: {cnbc_df.columns.tolist()}")
print(f"CNBC dataset missing values:\n{cnbc_df.isnull().sum()}")
```

```python
# Load Guardian news data
guardian_df = pd.read_csv("../data/raw/headline_data/guardian_headlines.csv")
guardian_df.head(10)
```

```python
print(f"Guardian dataset shape: {guardian_df.shape}")  
print(f"Guardian dataset columns: {guardian_df.columns.tolist()}")
print(f"Guardian dataset missing values:\n{guardian_df.isnull().sum()}")
```

```python
# Load Reuters news data
reuters_df = pd.read_csv("../data/raw/headline_data/reuters_headlines.csv")
reuters_df.head(10)
```

```python
print(f"Reuters dataset shape: {reuters_df.shape}")
print(f"Reuters dataset columns: {reuters_df.columns.tolist()}")
print(f"Reuters dataset missing values:\n{reuters_df.isnull().sum()}")
```

### Checking time format

```python
print("-------CNBC Time Column Sample-------")
print(cnbc_df["Time"].head())
print("-------Guardian Time Column Sample-------")
print(guardian_df["Time"].head())
print("-------Reuters Time Column Sample-------")
print(reuters_df["Time"].head())
```

#### Modifying time formats

```python
cnbc_df["Date"] = pd.to_datetime(
    cnbc_df["Time"]
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.replace(" ET ", " ", regex=False),
    format="mixed"
).dt.strftime("%Y-%m-%d")
print("-------CNBC Date Column Sample-------")
print(cnbc_df["Date"].head())
```

```python
guardian_df["Date"] = (
    pd.to_datetime(guardian_df["Time"])
    .dt.strftime("%Y-%m-%d")
)
print("-------Guardian Date Column Sample-------")
print(guardian_df["Date"].head())
```

```python
guardian_df.isnull().sum()
```

```python
guardian_df[guardian_df["Time"].str.strip().str.match(r"^[A-Za-z]{3}-\d{2}$", na=False)]
```

```python
reuters_df["Date"] = pd.to_datetime(
    reuters_df["Time"],
    format="%b %d %Y"
)
print("-------Reuters Date Column Sample-------")
print(reuters_df["Date"].head())
```

## Combining all dataframes

```python
for df, source in [
    (cnbc_df, "cnbc"),
    (guardian_df, "guardian"),
    (reuters_df, "reuters"),
]:
    df["Source"] = source

combined_df = pd.concat(
    [
        cnbc_df[["Headlines", "Date", "Source"]],
        guardian_df[["Headlines", "Date", "Source"]],
        reuters_df[["Headlines", "Date", "Source"]],
    ],
    ignore_index=True,
)

combined_df["Date"] = pd.to_datetime(combined_df["Date"], errors="coerce")
```

```python
combined_df
```

```python
# Save the combined DataFrame to a new CSV file
combined_df.to_csv("../data/processed/news_combined.csv", index=False)
```
